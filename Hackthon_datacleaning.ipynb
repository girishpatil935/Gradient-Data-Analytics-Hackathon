```python
# ============================================================
# CELL 1 — IMPORTS + LOCATE DATASET
# ============================================================

import pandas as pd
import numpy as np
import os

print("=" * 70)
print("PHASE 1 — INITIALIZATION")
print("=" * 70)

# Try the common Colab locations.
possible_paths = [
    "/content/Dataset.xlsx",
    "Dataset.xlsx",
    "/mnt/data/Dataset.xlsx"
]

DATASET_PATH = None

for path in possible_paths:
    if os.path.exists(path):
        DATASET_PATH = path
        break

if DATASET_PATH is None:
    raise FileNotFoundError(
        "Dataset.xlsx was not found in the current Colab runtime. "
        "If you already uploaded it to Colab, check the Files panel."
    )

print("Dataset found at:")
print(DATASET_PATH)

# Open workbook
xls = pd.ExcelFile(DATASET_PATH)

print("\nSheets available:")
for sheet in xls.sheet_names:
    print("-", sheet)
```

```python
# ============================================================
# CELL 2 — LOAD ALL REQUIRED SHEETS
# ============================================================

print("=" * 70)
print("LOADING DATASETS")
print("=" * 70)

orders = pd.read_excel(
    DATASET_PATH,
    sheet_name="orders"
)

order_items = pd.read_excel(
    DATASET_PATH,
    sheet_name="order_items"
)

order_payments = pd.read_excel(
    DATASET_PATH,
    sheet_name="order_payments"
)

order_reviews = pd.read_excel(
    DATASET_PATH,
    sheet_name="order_reviews"
)

customers = pd.read_excel(
    DATASET_PATH,
    sheet_name="customers"
)

products = pd.read_excel(
    DATASET_PATH,
    sheet_name="products"
)

category_translation = pd.read_excel(
    DATASET_PATH,
    sheet_name="category_translation"
)

print("Orders:", orders.shape)
print("Order items:", order_items.shape)
print("Payments:", order_payments.shape)
print("Reviews:", order_reviews.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
print("Category translation:", category_translation.shape)
```

```python
# ============================================================
# CELL 3 — COLUMN VALIDATION
# ============================================================

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "customers": customers,
    "products": products,
    "category_translation": category_translation
}

for name, df in datasets.items():

    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)

    print("Rows:", len(df))
    print("Columns:")

    for i, col in enumerate(df.columns, start=1):
        print(f"{i}. {col}")
```

```python
# ============================================================
# CELL 4 — CLEAN ORDER TIMESTAMPS
# ============================================================

print("=" * 70)
print("CLEANING ORDER TIMESTAMPS")
print("=" * 70)

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(
        orders[col],
        errors="coerce"
    )

print("Datetime conversion complete.")

print("\nMissing datetime values:")
print(orders[date_columns].isnull().sum())
```

```python
# ============================================================
# CELL 5 — ORDER TABLE VALIDATION
# ============================================================

print("=" * 70)
print("ORDERS VALIDATION")
print("=" * 70)

print("Rows:", len(orders))
print("Unique order IDs:", orders["order_id"].nunique())
print("Duplicate order IDs:", orders["order_id"].duplicated().sum())

assert orders["order_id"].nunique() == len(orders), \
    "Orders table contains duplicate order IDs."

print("\nOrder status distribution:")
print(orders["order_status"].value_counts(dropna=False))

print("\nCustomer ID missing:")
print(orders["customer_id"].isnull().sum())
```

```python
# ============================================================
# CELL 6 — ORDER-LEVEL ITEM SUMMARY
# ============================================================

print("=" * 70)
print("CREATING ORDER-LEVEL ITEM SUMMARY")
print("=" * 70)

# Make sure numeric columns are numeric
order_items["order_item_id"] = pd.to_numeric(
    order_items["order_item_id"],
    errors="coerce"
)

order_items["price"] = pd.to_numeric(
    order_items["price"],
    errors="coerce"
)

order_items["freight_value"] = pd.to_numeric(
    order_items["freight_value"],
    errors="coerce"
)

item_summary = (
    order_items
    .groupby("order_id")
    .agg(
        item_count=("order_item_id", "count"),
        unique_products=("product_id", "nunique"),
        unique_sellers=("seller_id", "nunique"),
        product_value=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
    .reset_index()
)

item_summary["order_value"] = (
    item_summary["product_value"] +
    item_summary["freight_value"]
)

print("Orders represented:", len(item_summary))
print("Unique order IDs:", item_summary["order_id"].nunique())
print("Duplicate order IDs:", item_summary["order_id"].duplicated().sum())

display(item_summary.head())
```

```python

print("=" * 70)
print("CREATING PAYMENT SUMMARY")
print("=" * 70)

order_payments["payment_value"] = pd.to_numeric(
    order_payments["payment_value"],
    errors="coerce"
)

order_payments["payment_installments"] = pd.to_numeric(
    order_payments["payment_installments"],
    errors="coerce"
)

payment_summary = (
    order_payments
    .groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_count=("payment_sequential", "count"),
        max_installments=("payment_installments", "max")
    )
    .reset_index()
)

print("Orders represented:", len(payment_summary))
print("Unique order IDs:", payment_summary["order_id"].nunique())
print("Duplicate order IDs:", payment_summary["order_id"].duplicated().sum())

# Special payment validation
max_payment_order = (
    payment_summary
    .sort_values("payment_count", ascending=False)
    .head(1)
)

print("\nHighest payment-count order:")
display(max_payment_order)
```

```python
print("=" * 70)
print("CREATING REVIEW SUMMARY")
print("=" * 70)

order_reviews["review_score"] = pd.to_numeric(
    order_reviews["review_score"],
    errors="coerce"
)

review_summary = (
    order_reviews
    .groupby("order_id")
    .agg(
        review_count=("review_id", "count"),
        review_score_mean=("review_score", "mean"),
        review_score_min=("review_score", "min"),
        review_score_max=("review_score", "max")
    )
    .reset_index()
)

print("Orders represented:", len(review_summary))
print("Unique order IDs:", review_summary["order_id"].nunique())
print("Duplicate order IDs:", review_summary["order_id"].duplicated().sum())

# Detect conflicting review scores
review_conflicts = (
    order_reviews
    .groupby("order_id")["review_score"]
    .nunique()
)

conflicting_orders = review_conflicts[
    review_conflicts > 1
]

print("\nOrders with conflicting review scores:",
      len(conflicting_orders))

print("\nFirst 20 conflicting orders:")
display(conflicting_orders.head(20))
```

```python
print("=" * 70)
print("BUILDING MASTER V1")
print("=" * 70)

# Start from orders because MASTER V1 is one row per order.
master_v1 = orders.copy()

# Merge item summary
master_v1 = master_v1.merge(
    item_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Merge payment summary
master_v1 = master_v1.merge(
    payment_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Merge review summary
master_v1 = master_v1.merge(
    review_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(master_v1))
print("Unique order IDs:", master_v1["order_id"].nunique())
print("Duplicate order IDs:",
      master_v1["order_id"].duplicated().sum())
```

```python
print("=" * 70)
print("CREATING DELIVERY FEATURES")
print("=" * 70)

master_v1["delivery_days"] = (
    master_v1["order_delivered_customer_date"]
    - master_v1["order_purchase_timestamp"]
).dt.total_seconds() / 86400

master_v1["estimated_delivery_days"] = (
    master_v1["order_estimated_delivery_date"]
    - master_v1["order_purchase_timestamp"]
).dt.total_seconds() / 86400

master_v1["delivery_delay_days"] = (
    master_v1["order_delivered_customer_date"]
    - master_v1["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Use pandas nullable integer type for the label
master_v1["delivery_performance"] = pd.Series(
    pd.NA,
    index=master_v1.index,
    dtype="string"
)

delivered_mask = (
    master_v1["order_delivered_customer_date"].notna()
)

master_v1.loc[
    delivered_mask &
    (master_v1["delivery_delay_days"] <= 0),
    "delivery_performance"
] = "On Time"

master_v1.loc[
    delivered_mask &
    (master_v1["delivery_delay_days"] > 0),
    "delivery_performance"
] = "Late"

print("Delivery performance:")
print(
    master_v1["delivery_performance"]
    .value_counts(dropna=False)
)
```

```python
print("=" * 70)
print("MASTER V1 VALIDATION")
print("=" * 70)

print("Rows:", len(master_v1))
print("Columns:", len(master_v1.columns))
print("Unique order IDs:", master_v1["order_id"].nunique())
print("Duplicate order IDs:",
      master_v1["order_id"].duplicated().sum())

assert len(master_v1) == 99441, \
    f"Expected 99,441 rows, got {len(master_v1)}"

assert master_v1["order_id"].nunique() == 99441, \
    "MASTER V1 does not contain 99,441 unique orders."

assert master_v1["order_id"].duplicated().sum() == 0, \
    "MASTER V1 contains duplicate order IDs."

print("\nMASTER V1 structure is valid.")

print("\nMissing values:")
print(
    master_v1.isnull()
    .sum()
    .sort_values(ascending=False)
    .head(20)
)
```

```python
print("=" * 70)
print("PREPARING PRODUCT DATA")
print("=" * 70)

# Keep only useful product columns
product_cols = [
    "product_id",
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

product_cols = [
    col for col in product_cols
    if col in products.columns
]

products_clean = products[product_cols].copy()

# Create product volume
if all(
    col in products_clean.columns
    for col in [
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ]
):

    products_clean["product_volume_cm3"] = (
        products_clean["product_length_cm"]
        * products_clean["product_height_cm"]
        * products_clean["product_width_cm"]
    )

# Clean translation table
category_translation = category_translation.copy()

print("Products:", len(products_clean))
print("Category translations:", len(category_translation))

print("\nProduct columns:")
print(products_clean.columns.tolist())

print("\nTranslation columns:")
print(category_translation.columns.tolist())
```

```python
print("=" * 70)
print("CREATING ITEM + PRODUCT DATASET")
print("=" * 70)

item_product = order_items.merge(
    products_clean,
    on="product_id",
    how="left"
)

print("Rows:", len(item_product))

print(
    "Missing product matches:",
    item_product["product_category_name"].isna().sum()
)

display(item_product.head())
```

```python
print("=" * 70)
print("ADDING ENGLISH CATEGORIES")
print("=" * 70)

# Standard Olist translation columns:
# product_category_name -> product_category_name_english

translation_cols = category_translation.columns.tolist()

print("Translation columns:", translation_cols)

# Rename only if standard columns exist
if (
    "product_category_name" in category_translation.columns
    and
    "product_category_name_english" in category_translation.columns
):

    category_translation_clean = (
        category_translation[
            [
                "product_category_name",
                "product_category_name_english"
            ]
        ]
        .drop_duplicates("product_category_name")
    )

    item_product = item_product.merge(
        category_translation_clean,
        on="product_category_name",
        how="left"
    )

else:
    raise ValueError(
        "Expected category translation columns were not found."
    )

print(
    "Missing English categories:",
    item_product["product_category_name_english"].isna().sum()
)

display(
    item_product[
        [
            "product_id",
            "product_category_name",
            "product_category_name_english"
        ]
    ].head(10)
)
```

```python
print("=" * 70)
print("ORDER-LEVEL PRODUCT & CATEGORY AGGREGATION")
print("=" * 70)

# Helper function to safely join unique categories
def join_unique_categories(series):
    values = (
        series
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    values = sorted(values)

    if len(values) == 0:
        return pd.NA

    return "|".join(values)


product_order_summary = (
    item_product
    .groupby("order_id")
    .agg(
        product_category_count=(
            "product_category_name_english",
            "nunique"
        ),

        primary_category=(
            "product_category_name_english",
            lambda x: (
                x.mode().iloc[0]
                if not x.mode().empty
                else pd.NA
            )
        ),

        product_categories=(
            "product_category_name_english",
            join_unique_categories
        ),

        avg_product_price=(
            "price",
            "mean"
        ),

        min_product_price=(
            "price",
            "min"
        ),

        max_product_price=(
            "price",
            "max"
        ),

        total_product_photos=(
            "product_photos_qty",
            "sum"
        ),

        avg_product_photos_qty=(
            "product_photos_qty",
            "mean"
        ),

        avg_product_weight_g=(
            "product_weight_g",
            "mean"
        ),

        avg_product_volume_cm3=(
            "product_volume_cm3",
            "mean"
        )
    )
    .reset_index()
)

product_order_summary["has_multiple_categories"] = (
    product_order_summary["product_category_count"] > 1
)

print("Orders represented:",
      len(product_order_summary))

print("Unique order IDs:",
      product_order_summary["order_id"].nunique())

print("Duplicate order IDs:",
      product_order_summary["order_id"].duplicated().sum())

print(
    "\nOrders with multiple categories:",
    product_order_summary["has_multiple_categories"]
    .sum()
)

display(product_order_summary.head())
```

```python
print("=" * 70)
print("ADDING PRODUCT FEATURES TO MASTER V1")
print("=" * 70)

master_v1 = master_v1.merge(
    product_order_summary,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(master_v1))
print("Unique orders:", master_v1["order_id"].nunique())
print("Duplicate orders:",
      master_v1["order_id"].duplicated().sum())

assert len(master_v1) == 99441
assert master_v1["order_id"].nunique() == 99441
assert master_v1["order_id"].duplicated().sum() == 0

print("\nProduct enrichment successful.")
```

```python
print("=" * 70)
print("SPEND-WEIGHTED CATEGORY FEATURES")
print("=" * 70)

# Create item-level category spending dataset
item_category_spend = item_product[
    [
        "order_id",
        "product_category_name_english",
        "price"
    ]
].copy()

item_category_spend["price"] = pd.to_numeric(
    item_category_spend["price"],
    errors="coerce"
)

print("Rows:", len(item_category_spend))

print(
    "Missing category:",
    item_category_spend[
        "product_category_name_english"
    ].isna().sum()
)

# Remove rows where category cannot be identified
category_spend_valid = (
    item_category_spend
    .dropna(subset=["product_category_name_english"])
)

# Aggregate by order + category
order_category_spend = (
    category_spend_valid
    .groupby(
        [
            "order_id",
            "product_category_name_english"
        ]
    )
    .agg(
        category_spend=("price", "sum")
    )
    .reset_index()
)

print(
    "\nOrder-category rows:",
    len(order_category_spend)
)
```

```python
print("=" * 70)
print("CALCULATING SPEND CONCENTRATION")
print("=" * 70)

# Total identified category spend
order_category_totals = (
    order_category_spend
    .groupby("order_id")
    ["category_spend"]
    .sum()
    .rename("category_identified_spend")
)

# Number of identified categories
order_category_counts = (
    order_category_spend
    .groupby("order_id")
    ["product_category_name_english"]
    .nunique()
    .rename("category_spend_count")
)

# Highest-spend category
top_category = (
    order_category_spend
    .sort_values(
        [
            "order_id",
            "category_spend"
        ],
        ascending=[True, False]
    )
    .drop_duplicates("order_id")
)

top_category = top_category[
    [
        "order_id",
        "product_category_name_english",
        "category_spend"
    ]
].rename(
    columns={
        "product_category_name_english":
            "top_spend_category",
        "category_spend":
            "top_category_spend"
    }
)

# Combine
spend_features = (
    order_category_totals
    .to_frame()
    .join(order_category_counts)
    .reset_index()
    .merge(
        top_category,
        on="order_id",
        how="left",
        validate="one_to_one"
    )
)

# Top category share
spend_features["top_category_spend_share"] = (
    spend_features["top_category_spend"]
    /
    spend_features["category_identified_spend"]
)

# Herfindahl-style category concentration:
# sum of squared category spending shares
order_category_spend["spend_share"] = (
    order_category_spend["category_spend"]
    /
    order_category_spend
    .groupby("order_id")["category_spend"]
    .transform("sum")
)

concentration = (
    order_category_spend
    .assign(
        squared_share=lambda x:
        x["spend_share"] ** 2
    )
    .groupby("order_id")["squared_share"]
    .sum()
    .rename("category_spend_concentration")
)

spend_features = spend_features.merge(
    concentration,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Orders represented:",
      len(spend_features))

print("Duplicate order IDs:",
      spend_features["order_id"].duplicated().sum())

display(spend_features.head(10))
```

```python
print("=" * 70)
print("ADDING SPEND FEATURES")
print("=" * 70)

master_v1 = master_v1.merge(
    spend_features,
    on="order_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(master_v1))
print("Unique orders:", master_v1["order_id"].nunique())
print("Duplicate orders:",
      master_v1["order_id"].duplicated().sum())

assert len(master_v1) == 99441
assert master_v1["order_id"].nunique() == 99441
assert master_v1["order_id"].duplicated().sum() == 0

print("\nSpend enrichment successful.")
```

```python
print("=" * 70)
print("PREPARING CUSTOMER INFORMATION")
print("=" * 70)

customer_columns = [
    "customer_id",
    "customer_unique_id"
]

customer_clean = customers[
    [
        col for col in customer_columns
        if col in customers.columns
    ]
].copy()

# Ensure customer_id is unique before merging
customer_clean = (
    customer_clean
    .drop_duplicates("customer_id")
)

master_v1 = master_v1.merge(
    customer_clean,
    on="customer_id",
    how="left",
    validate="one_to_one"
)

print("Rows:", len(master_v1))
print("Unique orders:", master_v1["order_id"].nunique())
print("Duplicate orders:",
      master_v1["order_id"].duplicated().sum())

print(
    "\nMissing customer_unique_id:",
    master_v1["customer_unique_id"].isna().sum()
)

assert len(master_v1) == 99441
assert master_v1["order_id"].duplicated().sum() == 0
```

```python
print("=" * 70)
print("CREATING CUSTOMER-LEVEL FEATURES")
print("=" * 70)

# Only customers with a valid customer_unique_id
customer_valid = master_v1[
    master_v1["customer_unique_id"].notna()
].copy()

print(
    "Orders with valid customer_unique_id:",
    len(customer_valid)
)

print(
    "Orders without customer_unique_id:",
    len(master_v1) - len(customer_valid)
)

# ------------------------------------------------------------
# Basic customer behavior
# ------------------------------------------------------------

customer_features = (
    customer_valid
    .groupby("customer_unique_id")
    .agg(
        customer_order_count=(
            "order_id",
            "nunique"
        ),

        customer_total_spend=(
            "order_value",
            "sum"
        ),

        customer_avg_order_value=(
            "order_value",
            "mean"
        ),

        customer_avg_review_score=(
            "review_score_mean",
            "mean"
        ),

        customer_late_order_rate=(
            "delivery_performance",
            lambda x: (
                (x == "Late").sum()
                /
                x.notna().sum()
                if x.notna().sum() > 0
                else np.nan
            )
        ),

        customer_avg_delivery_delay=(
            "delivery_delay_days",
            "mean"
        ),

        customer_category_diversity=(
            "primary_category",
            "nunique"
        )
    )
    .reset_index()
)

# Repeat purchase indicator
customer_features[
    "customer_repeat_purchase_indicator"
] = (
    customer_features["customer_order_count"] > 1
).astype("int64")

print("Customer records created:",
      len(customer_features))

display(customer_features.head(10))
```

```python
print("=" * 70)
print("MERGING CUSTOMER FEATURES")
print("=" * 70)

master_v1 = master_v1.merge(
    customer_features,
    on="customer_unique_id",
    how="left",
    validate="many_to_one"
)

print("Rows:", len(master_v1))
print("Unique order IDs:",
      master_v1["order_id"].nunique())

print("Duplicate order IDs:",
      master_v1["order_id"].duplicated().sum())

assert len(master_v1) == 99441
assert master_v1["order_id"].nunique() == 99441
assert master_v1["order_id"].duplicated().sum() == 0

print("\nCustomer enrichment successful.")
```

```python
print("=" * 70)
print("PRODUCT, CATEGORY & SPEND CONSISTENCY CHECK")
print("=" * 70)

# ------------------------------------------------------------
# 1. Product/category feature availability
# ------------------------------------------------------------

print("\n[1] PRODUCT/CATEGORY FEATURE COVERAGE")
print("-" * 70)

product_cols = [
    "item_count",
    "unique_products",
    "unique_sellers",
    "product_value",
    "freight_value",
    "order_value",
    "product_category_count",
    "primary_category",
    "avg_product_price"
]

for col in product_cols:
    missing = master_v1[col].isna().sum()
    available = master_v1[col].notna().sum()

    print(
        f"{col:35} | "
        f"Available: {available:6} | "
        f"Missing: {missing:6}"
    )


# ------------------------------------------------------------
# 2. Product count consistency
# ------------------------------------------------------------

print("\n[2] PRODUCT COUNT CONSISTENCY")
print("-" * 70)

product_count_issue = (
    master_v1["unique_products"].notna() &
    master_v1["item_count"].notna() &
    (master_v1["unique_products"] > master_v1["item_count"])
).sum()

print(
    "Orders where unique_products > item_count:",
    product_count_issue
)


# ------------------------------------------------------------
# 3. Seller count consistency
# ------------------------------------------------------------

print("\n[3] SELLER COUNT CONSISTENCY")
print("-" * 70)

seller_count_issue = (
    master_v1["unique_sellers"].notna() &
    master_v1["item_count"].notna() &
    (master_v1["unique_sellers"] > master_v1["item_count"])
).sum()

print(
    "Orders where unique_sellers > item_count:",
    seller_count_issue
)


# ------------------------------------------------------------
# 4. Category count consistency
# ------------------------------------------------------------

print("\n[4] CATEGORY COUNT CONSISTENCY")
print("-" * 70)

category_count_issue = (
    master_v1["product_category_count"].notna() &
    master_v1["item_count"].notna() &
    (
        master_v1["product_category_count"]
        > master_v1["item_count"]
    )
).sum()

print(
    "Orders where category count > item count:",
    category_count_issue
)


# ------------------------------------------------------------
# 5. Spend consistency
# ------------------------------------------------------------

print("\n[5] CATEGORY SPEND CONSISTENCY")
print("-" * 70)

spend_issue = (
    master_v1["category_identified_spend"].notna() &
    master_v1["order_value"].notna() &
    (
        master_v1["category_identified_spend"]
        > master_v1["order_value"] + 0.01
    )
).sum()

print(
    "Orders where identified category spend > order value:",
    spend_issue
)


# ------------------------------------------------------------
# 6. Top category spend share
# ------------------------------------------------------------

print("\n[6] TOP CATEGORY SPEND SHARE")
print("-" * 70)

share = master_v1["top_category_spend_share"].dropna()

invalid_share = (
    (share < 0) |
    (share > 1)
).sum()

print("Minimum share:", share.min())
print("Maximum share:", share.max())
print("Invalid shares:", invalid_share)


# ------------------------------------------------------------
# 7. Multiple category indicator
# ------------------------------------------------------------

print("\n[7] MULTIPLE CATEGORY VALIDATION")
print("-" * 70)

multi_category = master_v1[
    master_v1["has_multiple_categories"].notna() &
    master_v1["product_category_count"].notna()
].copy()

multiple_category_issue = (
    (
        multi_category["has_multiple_categories"] == True
    ) &
    (
        multi_category["product_category_count"] <= 1
    )
).sum()

single_category_issue = (
    (
        multi_category["has_multiple_categories"] == False
    ) &
    (
        multi_category["product_category_count"] > 1
    )
).sum()

print(
    "Multiple-category flag = True but category count <= 1:",
    multiple_category_issue
)

print(
    "Multiple-category flag = False but category count > 1:",
    single_category_issue
)


# ------------------------------------------------------------
# FINAL RESULT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCT & SPEND CONSISTENCY RESULT")
print("=" * 70)

total_issues = (
    product_count_issue
    + seller_count_issue
    + category_count_issue
    + spend_issue
    + invalid_share
    + multiple_category_issue
    + single_category_issue
)

if total_issues == 0:
    print("✓ PASS — Product/category/spend features are consistent.")
else:
    print(
        "⚠ REVIEW —",
        total_issues,
        "potential product/category/spend issues found."
    )
```

```python
customer_features = [
    "customer_unique_id",
    "customer_order_count",
    "customer_total_spend",
    "customer_avg_order_value",
    "customer_avg_review_score",
    "customer_late_order_rate",
    "customer_avg_delivery_delay",
    "customer_category_diversity",
    "customer_repeat_purchase_indicator"
]

print("=" * 70)
print("CUSTOMER FEATURE COVERAGE")
print("=" * 70)

for col in customer_features:
    if col in master_v1.columns:
        available = master_v1[col].notna().sum()
        missing = master_v1[col].isna().sum()
        print(f"{col:40} | Available: {available:6,} | Missing: {missing:6,}")
    else:
        print(f"{col:40} | ❌ COLUMN MISSING")
```

```python
print("=" * 70)
print("MISSING CUSTOMER ID CHECK")
print("=" * 70)

missing_customer_id = master_v1["customer_unique_id"].isna()

print(f"Orders with customer_unique_id:     {(~missing_customer_id).sum():,}")
print(f"Orders missing customer_unique_id:  {missing_customer_id.sum():,}")

print("\nCustomer features on orders with missing customer ID:")

customer_feature_only = [
    "customer_order_count",
    "customer_total_spend",
    "customer_avg_order_value",
    "customer_avg_review_score",
    "customer_late_order_rate",
    "customer_avg_delivery_delay",
    "customer_category_diversity",
    "customer_repeat_purchase_indicator"
]

for col in customer_feature_only:
    count = master_v1.loc[missing_customer_id, col].notna().sum()
    print(f"{col:40} | Non-missing values: {count:,}")
```

```python
print("=" * 70)
print("REPEAT PURCHASE VALIDATION")
print("=" * 70)

valid_customer = master_v1["customer_unique_id"].notna()

expected_repeat = (
    master_v1.loc[valid_customer, "customer_order_count"] > 1
).astype("Int64")

actual_repeat = master_v1.loc[
    valid_customer, "customer_repeat_purchase_indicator"
]

repeat_mismatch = (
    expected_repeat != actual_repeat
).sum()

print(f"Valid customer orders: {valid_customer.sum():,}")
print(f"Repeat indicator mismatches: {repeat_mismatch:,}")

if repeat_mismatch == 0:
    print("✓ PASS — Repeat purchase indicator is consistent.")
else:
    print("⚠ REVIEW — Repeat purchase indicator has inconsistencies.")
```

```python
print("=" * 70)
print("CUSTOMER ORDER COUNT CONSISTENCY")
print("=" * 70)

customer_count_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_order_count"]
    .nunique()
)

inconsistent_customers = (customer_count_check > 1).sum()

print(f"Customers checked: {len(customer_count_check):,}")
print(f"Customers with inconsistent order counts: {inconsistent_customers:,}")

if inconsistent_customers == 0:
    print("✓ PASS — Customer order counts are consistent.")
else:
    print("⚠ REVIEW — Some customers have inconsistent order counts.")
```

```python
print("=" * 70)
print("CUSTOMER TOTAL SPEND CONSISTENCY")
print("=" * 70)

spend_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_total_spend"]
    .nunique()
)

inconsistent_spend = (spend_check > 1).sum()

print(f"Customers checked: {len(spend_check):,}")
print(f"Customers with inconsistent total spend: {inconsistent_spend:,}")

if inconsistent_spend == 0:
    print("✓ PASS — Customer total spend is consistent.")
else:
    print("⚠ REVIEW — Some customers have inconsistent total spend.")
```

```python
print("=" * 70)
print("CUSTOMER AVERAGE ORDER VALUE CONSISTENCY")
print("=" * 70)

avg_order_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_avg_order_value"]
    .nunique()
)

inconsistent_avg_order = (avg_order_check > 1).sum()

print(f"Customers checked: {len(avg_order_check):,}")
print(
    f"Customers with inconsistent average order value: "
    f"{inconsistent_avg_order:,}"
)

if inconsistent_avg_order == 0:
    print("✓ PASS — Customer average order value is consistent.")
else:
    print("⚠ REVIEW — Some customers have inconsistent average order value.")
```

```python
print("=" * 70)
print("CUSTOMER REVIEW SCORE CONSISTENCY")
print("=" * 70)

review_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_avg_review_score"]
    .nunique()
)

inconsistent_review = (review_check > 1).sum()

print(f"Customers checked: {len(review_check):,}")
print(
    f"Customers with inconsistent average review score: "
    f"{inconsistent_review:,}"
)

if inconsistent_review == 0:
    print("✓ PASS — Customer review scores are consistent.")
else:
    print("⚠ REVIEW — Some customers have inconsistent review scores.")
```

```python
print("=" * 70)
print("CUSTOMER DELIVERY FEATURE CONSISTENCY")
print("=" * 70)

late_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_late_order_rate"]
    .nunique()
)

delay_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_avg_delivery_delay"]
    .nunique()
)

inconsistent_late = (late_check > 1).sum()
inconsistent_delay = (delay_check > 1).sum()

print(
    f"Customers with inconsistent late-order rate: "
    f"{inconsistent_late:,}"
)

print(
    f"Customers with inconsistent average delivery delay: "
    f"{inconsistent_delay:,}"
)

if inconsistent_late == 0:
    print("✓ Late-order rate: PASS")
else:
    print("⚠ Late-order rate: REVIEW")

if inconsistent_delay == 0:
    print("✓ Average delivery delay: PASS")
else:
    print("⚠ Average delivery delay: REVIEW")
```

```python
print("=" * 70)
print("CUSTOMER CATEGORY DIVERSITY CONSISTENCY")
print("=" * 70)

category_check = (
    master_v1[valid_customer]
    .groupby("customer_unique_id")["customer_category_diversity"]
    .nunique()
)

inconsistent_category = (category_check > 1).sum()

print(f"Customers checked: {len(category_check):,}")
print(
    f"Customers with inconsistent category diversity: "
    f"{inconsistent_category:,}"
)

if inconsistent_category == 0:
    print("✓ PASS — Customer category diversity is consistent.")
else:
    print("⚠ REVIEW — Some customers have inconsistent category diversity.")
```

```python
print("=" * 70)
print("CUSTOMER FEATURE LOGICAL VALIDATION")
print("=" * 70)

valid = master_v1["customer_unique_id"].notna()

checks = {}

checks["order_count_less_than_1"] = (
    master_v1.loc[valid, "customer_order_count"] < 1
).sum()

checks["negative_total_spend"] = (
    master_v1.loc[valid, "customer_total_spend"] < 0
).sum()

checks["negative_avg_order_value"] = (
    master_v1.loc[valid, "customer_avg_order_value"] < 0
).sum()

checks["review_score_below_1"] = (
    master_v1.loc[valid, "customer_avg_review_score"] < 1
).sum()

checks["review_score_above_5"] = (
    master_v1.loc[valid, "customer_avg_review_score"] > 5
).sum()

checks["late_rate_below_0"] = (
    master_v1.loc[valid, "customer_late_order_rate"] < 0
).sum()

checks["late_rate_above_1"] = (
    master_v1.loc[valid, "customer_late_order_rate"] > 1
).sum()

checks["category_diversity_below_0"] = (
    master_v1.loc[valid, "customer_category_diversity"] < 0
).sum()

for name, count in checks.items():
    print(f"{name:40} | {count:,}")

total_issues = sum(checks.values())

print("\n" + "-" * 70)

if total_issues == 0:
    print("✓ PASS — No logical customer-feature violations found.")
else:
    print(f"⚠ REVIEW — {total_issues:,} logical issues found.")
```
